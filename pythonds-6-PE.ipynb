```python
"""
1. Set up a random experiment to test the difference between a sequential search and
a binary search on a list of integers
"""
"""
Process thinking:
main functions:
- a generator of sorted list of integer number between 0, 1000000
- Timer
- sequential_search
- binary_search
"""
import random, time

def time_it(func):
    def wrapper(*args, **kargs):
        start_time = time.perf_counter_ns()
        result = func(*args, **kargs)
        end_time = time.perf_counter_ns()
        elapsed_time = end_time - start_time
        print(f"==> Function '{func.__name__}' took {elapsed_time} nanoseconds to execute.")
        return result
    return wrapper

def generator_list(size_of_list: int) -> list:
    res: list[int]= []
    for num in range(size_of_list):
        res.append(random.randint(0,1000000))
    res.sort()
    return res
@time_it
def sequential_search(num_found: int, list_of_numbers: list[int])-> bool:
    for num in list_of_numbers:
        if num_found == num:
            return True
    return False

@time_it
def binary_search(num_found: int, list_of_numbers: list[int]) -> bool:
    length_of_list: int = len(list_of_numbers)
    new_len: list = list_of_numbers
    low: int = 0
    high: int = length_of_list
    while low < high:
        mid: int = (low + high)//2
        if list_of_numbers[mid] == num_found:
            return True
        elif list_of_numbers[mid] < num_found:
            low = mid+1
        else:
            high = mid
    return False

def simulator(num_of_simulations: int, num_of_elements: int) -> None:
    nums: list[int] = generator_list(num_of_elements)

    for i in range(num_of_simulations):
        num_to_search: int = random.randint(0, 1000000)
        print(f"num being searched: {num_to_search}")
        sequential_bool: bool = sequential_search(num_to_search, nums)
        binary_bool: bool = binary_search(num_to_search, nums)
        if (sequential_bool and binary_bool):
            print(f"num: {num_to_search} was found")
        else:
            print(f"num: {num_to_search} UNFOUND")

simulator(5, 10000)
```

```python
"""
2. Use the binary search functions given in the text(recursive and iterative). 
Generate a random, ordered list of integers and do a benchmark analysis for each one.
what are you result? Can you explain them?

Result: the binary search iterative performs better than a binary search recursive.
Explain: it is due to space time where the iterative tends to have a constant space time, 
the binary search recursive will tend to increase due to the creation of sub-function(stack) for each new slice to solve the sub-problems.
"""
import random, timeit

def generator_list(size_of_list: int) -> list:
    res: list[int]= []
    for num in range(size_of_list):
        res.append(random.randint(0,1000000))
    res.sort()
    return res

def binarySearchRecursive(alist, item):
    if len(alist) == 0:
        return False
    else:
        midpoint = len(alist)//2
        if alist[midpoint]==item:
          return True
        else:
          if item<alist[midpoint]:
            return binarySearchRecursive(alist[:midpoint],item)
          else:
             return binarySearchRecursive(alist[midpoint+1:], item)


def binarySearchIterative(alist, item):
    first = 0
    last = len(alist)-1
    found = False

    while first <= last and not found:
        midpoint = (first + last)//2
        if alist[midpoint] == item:
           found = True
        else:
            if item < alist[midpoint]:
               last = midpoint-1
            else:
               first = midpoint+1
    return found

def printSeconds(datas: list):
   ind: int = 0
   for data in datas:
    ind +=1
    print(f"{ind}: it took {data} seconds")
    
data_list: list = generator_list(10000)
numFound = random.randint(0,1000000)

TEST_CODE_RECURSIVE = '''
binarySearchRecursive(data_list, numFound)

''' 
TEST_CODE_ITERATIVE = ''' 
binarySearchIterative(data_list, numFound)
'''

times_recursive = timeit.repeat(stmt=TEST_CODE_RECURSIVE, globals=globals(), repeat=10, number=10000)
times_iterative = timeit.repeat(stmt=TEST_CODE_ITERATIVE, globals=globals(), repeat=10, number=10000)

print("# Recursive Function")
print()
printSeconds(times_recursive)
print()

print("# Iterative Function")
print()
printSeconds(times_iterative)
 
```

```python
"""
4. implement the len method (__len__) for the hash table Map ADT implementation.
"""
class HashTable:
    def __init__(self):
        self.size = 11
        self.slots = [None] * self.size
        self.data = [None] * self.size

    # hashfunction is the remainder of the key divided by size
    def hashfunction(self, key, size):
        return key%size

    def rehash(self, oldhash, size):
        return (oldhash+1)%size

    def put(self, key, val):
        hashvalue = self.hashfunction(key, len(self.slots))
        if self.slots[hashvalue] == None:
            self.slots[hashvalue] = key
            self.data[hashvalue] = val
        else:
            if self.slots[hashvalue] == key:
                self.data[hashvalue] = val
            else:
                nextslots = self.rehash(hashvalue, len(self.slots))
                while self.slots[nextslots] != None and self.slots[nextslots] != key:
                    nextslots = self.rehash(nextslots, len(self.slots))
                if self.slots[nextslots] == None:
                    self.slots[nextslots] = key
                    self.data[nextslots] = val
                else:
                    self.data[nextslots] = val # replacement

    def get(self, key):
        startslot = self.hashfunction(key, len(self.size))
        found = False
        stop = False
        position = startslot
        data = None

        while self.slots[position] != None and not found and not stop:
            if self.slots[position] == key:
                found = True
                data = self.data[position]
            else:
                position = self.rehash(position, len(self.slots))
                if position == startslot:
                    stop = True
        return data
    
    def length_HT(self):
        count = 0
        for slot in self.slots:
            if slot != None:
                count +=1 
        return count

    def __getitem__(self, key):
        return self.get(key)
    
    def __setitem__(self, key, value):
        return self.put(key, value)
    
    def __len__(self):
        return self.length_HT()

H = HashTable()

            
H[54]="cat"

H[26]="dog"

H[93]="lion"

H[17]="tiger"

H[77]="bird"

H[31]="cow"

H[44]="goat"

H[55]="pig"

H[20]="chicken"
H.slots
len(H)
#H.data
```

```python
"""
6. How can you delete items from a hash table that uses chaining for collision resolution?
How about if open addressing is used? What are the special circumstances that must be handled?
Implement the del method for the HashTable class.
# Load Factor: as you delete and add items, the number of sentinels grows, This makes 'get' slower,  A common practice is to rehash the entire table(remove sentinels) once the load factor get too high
# Clustering: Linear probing creates clusters. Deletion without sentinels breaks the chain of the probes
"""

class HashTable:
    def __init__(self):
        self.size = 11
        self.slots = [None] * self.size
        self.data = [None] * self.size
        self.Delete = "__DELETED__" 

    # hashfunction is the remainder of the key divided by size
    def hashfunction(self, key, size):
        return key%size

    def rehash(self, oldhash, size):
        return (oldhash+1)%size

    def put(self, key, val):
        hashvalue = self.hashfunction(key, len(self.slots))
        if self.slots[hashvalue] == None:
            self.slots[hashvalue] = key
            self.data[hashvalue] = val
        else:
            if self.slots[hashvalue] == key:
                self.data[hashvalue] = val
            else:
                nextslots = self.rehash(hashvalue, len(self.slots))
                while self.slots[nextslots] != None and self.slots[nextslots] != key:
                    nextslots = self.rehash(nextslots, len(self.slots))
                if self.slots[nextslots] == None:
                    self.slots[nextslots] = key
                    self.data[nextslots] = val
                else:
                    self.data[nextslots] = val # replacement

    def get(self, key):
        startslot = self.hashfunction(key, len(self.slots))
        found = False
        stop = False
        position = startslot
        data = None

        while self.slots[position] != None and not found and not stop:
            if self.slots[position] == key:
                found = True
                data = self.data[position]
            else:
                position = self.rehash(position, len(self.slots))
                if position == startslot:
                    stop = True
        return data

    def length_HT(self):
        count = 0
        for slot in self.slots:
            if slot != None and slot != self.Delete:
                count +=1 
        return count

    def contains(self, key):
        startslot = self.hashfunction(key, len(self.slots))
        position = startslot
        while self.slots[position] != None: 
            if self.slots[position] == key:
                return True
            else:
                position = self.rehash(position, len(self.slots))
                if position == startslot:
                    return False
    
    # Delete the first occurence of the key only
    def delete(self, key):
        startslot = self.hashfunction(key, len(self.slots))
        position = startslot
        deleted = False
        stop = False

        while not deleted and not stop:
            if self.slots[position] == key:
                self.slots[position] = self.Delete
                self.data[position] = self.Delete
                deleted = True
            else:
                position = self.rehash(position, len(self.slots))
                if position == startslot:
                    stop = True
        if stop:
            return None


    def __getitem__(self, key):
        return self.get(key)
    
    def __setitem__(self, key, value):
        return self.put(key, value)
    
    def __len__(self):
        return self.length_HT()
    
    def __contains__(self, key):
        return self.contains(key)
    
    def __delitem__(self, key):
        return self.delete(key)

H = HashTable()

            
H[54]="cat"

H[26]="dog"

H[93]="lion"

H[17]="tiger"

H[77]="bird"

H[31]="cow"

H[44]="goat"

H[55]="pig"

H[20]="chicken"
print(H.slots)
print(len(H))
print(H[55])
55 in H
del H[44]
del H[77]
print(H.slots)
print(H.data)
print(H[55])
H[77]="Mouse"
print(H.slots)
print(H.data)
print(len(H))
```

```python
"""
7. In the hash table map implementation, the hash table size was chosen to be 101. if the table get full, 
this needs to be increased. Re-implement the 'put' method so that the table will automatically resize itself
when the loading factor reaches a predetermined value (you can decide the value based on your assessment of load
versus performance)
loading factor: Present the ratio between the number of items over tablesize
"""

class HashTable:
    def __init__(self):
        self.size = 7
        self.slots = [None] * self.size
        self.data = [None] * self.size
        self.DELETED = "__DELETED__"
        self.number_of_items = 0
        self.load_factor = 0.9
    
    def hashfunction(self, key, size):
        return key % size

    def rehash(self, oldkey, size):
        return (oldkey+1)%size

    def get(self, key):
        startslot = self.hashfunction(key, len(self.slots))
        position= startslot
        found = False
        stop = False
        data = None
        while not stop and not found:
            if self.slots[position] == key:
                data = self.data[position]  
                found = True
            else:
                position = self.rehash(position, len(self.slots))
                if position == startslot:
                    stop = True
        return data

    def put(self, key, value):
        if ((self.number_of_items + 1) / self.size) >= self.load_factor:
            print(f"hit the load refactor{self.number_of_items} and {self.size} and {self.load_factor}")
            self.reload_new_table(11) 
        hashvalue = self.hashfunction(key, len(self.slots))
        if self.slots[hashvalue] == None:
            self.slots[hashvalue] = key
            self.data[hashvalue] = value
        else:
            if self.slots[hashvalue] == key:
                self.data[hashvalue] = value
            else:
                nexthash = self.rehash(hashvalue, len(self.slots))
                while self.slots[nexthash] != None and self.slots[nexthash] != key:
                    nexthash = self.rehash(nexthash, len(self.slots))
                if self.slots[nexthash] == None:
                    self.slots[nexthash] = key
                    self.data[nexthash] = value
                else:
                    self.data[nexthash] = value
        self.number_of_items += 1
            

    def reload_new_table(self, new_size):
        old_slots = self.slots
        old_data = self.data

        new_slots = [None] * new_size
        new_data = [None] * new_size

        self.number_of_items = 0
        self.size = new_size
        self.slots = new_slots
        self.data = new_data
        for i in range(len(old_slots)):
            if old_slots[i] is not None and old_slots[i] != self.DELETED:
                self.put(old_slots[i], old_data[i])
        

    def __getitem__(self, key):
        self.get(key)
    
    def __setitem__(self, key, value):
        self.put(key, value)
```
